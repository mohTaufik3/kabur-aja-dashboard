# Make Review Dataset

Notebook ini bertujuan mengidentifikasi data dengan tingkat kepercayaan (confidence score) rendah dari hasil klasifikasi sentimen IndoBERT.

Data dengan confidence < 70% akan direview dan dikoreksi secara manual sebelum digunakan pada proses fine-tuning model.

```python
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv("/content/drive/MyDrive/KaburAjaDulu/data/sentiment_results.csv")
```

```python
low_conf = df[df["confidence"] < 70].copy()
high_conf = df[df["confidence"] >= 70].copy()

print(f"Confidence >= 70% : {len(high_conf)}")
print(f"Confidence < 70%  : {len(low_conf)}")
```

```python
high_conf["label_final"] = high_conf["sentiment"]
```

```python
low_conf["label_final"] = low_conf["sentiment"]

low_conf = (
    low_conf
    .sort_values("confidence", ascending=True)
    .reset_index(drop=True)
)
```

```python
low_conf["sentiment"].value_counts()
```

```python
(
    low_conf["sentiment"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
low_conf[[
        "full_text",
        "clean_text",
        "sentiment",
        "confidence",
        "label_final"
    ]].to_csv("review_low_confidence.csv",index=False)
```

```python
high_conf[[
        "full_text",
        "clean_text",
        "sentiment",
        "confidence",
        "label_final",
        "platform"
    ]].to_csv("high_confidence_data.csv",index=False)
```
