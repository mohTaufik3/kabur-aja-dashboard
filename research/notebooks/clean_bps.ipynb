```python
import pandas as pd
import os
from google.colab import drive
drive.mount('/content/drive')
```

```python
BASE_DIR = "/content/drive/MyDrive/KaburAjaDulu"
BPS_DIR = f"{BASE_DIR}/data/bps"
OUTPUT_DIR = f"{BPS_DIR}/clean_bps"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("BPS Directory :", BPS_DIR)
print("Output Directory :", OUTPUT_DIR)
```

```python
def load_csv(path):
    try:
        return pd.read_csv(path, encoding="utf-8")
    except:
        return pd.read_csv(path, encoding="latin1")
```

# Rata-rata Upah/Gaji

```python
for year in [2024, 2025]:
    print(f"\n{year}")
    path = (
        f"{BPS_DIR}/Rata-Rata Gaji/"
        f"Rata-Rata Upah_Gaji, {year}.csv"
        )

    df = load_csv(path)
    df_clean = df.iloc[3:].copy()
    df_clean.columns = [
        "Sektor_Ekonomi",
        "Februari",
        "Agustus",
        "Tahunan"
    ]

    df_clean = df_clean.dropna(subset=["Sektor_Ekonomi"])

    for col in [
        "Februari",
        "Agustus",
        "Tahunan"
    ]:

        df_clean[col] = pd.to_numeric(
            df_clean[col]
            .astype(str)
            .str.replace(",", "")
            .str.strip(),
            errors="coerce"
        )

    df_clean.to_csv(
        f"{OUTPUT_DIR}/gaji_{year}.csv",
        index=False
    )
    print(df_clean.head())
```

# Tingkat Pengangguran Terbuka (TPT)

```python
for year in [2024, 2025]:

    print(f"\n{year}")

    path = (
        f"{BPS_DIR}/TPT - Provinsi/"
        f"Tingkat Pengangguran Terbuka Menurut Provinsi, {year}.csv")

    df = load_csv(path)
    df_clean = df.iloc[3:].copy()
    df_clean.columns = [
        "Provinsi",
        "Februari",
        "Agustus",
        "Tahunan"
    ]
    df_clean = df_clean.dropna(subset=["Provinsi"])

    for col in [
        "Februari",
        "Agustus",
        "Tahunan"
    ]:
        df_clean[col] = pd.to_numeric(
            df_clean[col],
            errors="coerce"
        )

    df_clean.to_csv(
        f"{OUTPUT_DIR}/tpt_{year}.csv",
        index=False
    )
    print(df_clean.head())
```

```python
for year in [2024, 2025]:

    print(f"\n{year}")

    path = (
        f"{BPS_DIR}/Angkatan Kerja - Golongan Umur/"
        f"Angkatan Kerja (AK) Menurut Golongan Umur, {year}.csv"
    )

    df = load_csv(path)

    df_clean = df.iloc[3:].copy()

    df_clean.columns = [
        'Golongan_Umur',
        'Bekerja',
        'Unnamed_1',
        'Unnamed_2',
        'Pengangguran',
        'Unnamed_4',
        'Unnamed_5',
        'Total_AK',
        'Unnamed_7',
        'Unnamed_8',
        'Persen_Bekerja',
        'Unnamed_10',
        'Unnamed_11'
    ]

    df_clean = df_clean[
        [
            'Golongan_Umur',
            'Bekerja',
            'Pengangguran',
            'Total_AK',
            'Persen_Bekerja'
        ]
    ]

    df_clean = df_clean.dropna(
        subset=['Golongan_Umur']
    )

    df_clean.to_csv(
        f"{OUTPUT_DIR}/angkatan_kerja_umur_{year}.csv",
        index=False
    )

    print(df_clean.head())
```

```python
for year in [2024, 2025]:

    print(f"\n{year}")

    path = (
        f"{BPS_DIR}/BAK-UMUR/"
        f"Bukan Angkatan Kerja (BAK) Menurut Golongan Umur, {year}.csv"
    )

    df = load_csv(path)

    df_clean = df.iloc[3:].copy()

    df_clean.columns = [
        'Golongan_Umur',
        'Sekolah',
        'Unnamed_1',
        'Unnamed_2',
        'Urus_Rumah_Tangga',
        'Unnamed_4',
        'Unnamed_5',
        'Lainnya',
        'Unnamed_7',
        'Unnamed_8',
        'Total_BAK',
        'Unnamed_10',
        'Unnamed_11'
    ]

    df_clean = df_clean[
        [
            'Golongan_Umur',
            'Sekolah',
            'Urus_Rumah_Tangga',
            'Lainnya',
            'Total_BAK'
        ]
    ]

    df_clean = df_clean.dropna(
        subset=['Golongan_Umur']
    )

    df_clean.to_csv(
        f"{OUTPUT_DIR}/bukan_angkatan_kerja_umur_{year}.csv",
        index=False
    )

    print(df_clean.head())
```

# Penduduk 15 Tahun Ke Atas

```python
for year in [2024, 2025]:

    path = (
        f"{BPS_DIR}/Jumlah Penduduk/"
        f"Jumlah Penduduk Usia 15 tahun ke Atas Menurut Golongan Umur, {year}.csv"
    )

    df = load_csv(path)

    df_clean = df.iloc[3:].copy()

    df_clean.columns = [
        "Golongan_Umur",
        "Februari",
        "Agustus",
        "Tahunan"
    ]

    df_clean = df_clean.dropna(
        subset=["Golongan_Umur"]
    )

    output_path = (
        f"{OUTPUT_DIR}/penduduk_15_keatas_{year}.csv"
    )

    df_clean.to_csv(
        output_path,
        index=False
    )

    print(df_clean.head())
```

# Lapangan Pekerjaan Utama

```python
for year in [2024, 2025]:

    path = (
        f"{BPS_DIR}/Jumlah Penduduk/"
        f"Penduduk 15 Tahun Ke Atas yang Bekerja menurut Lapangan Pekerjaan Utama, {year}.csv"
    )

    df = load_csv(path)

    df_clean = df.iloc[3:].copy()

    df_clean.columns = [
        "Lapangan_Pekerjaan",
        "Februari",
        "Agustus"
    ]

    df_clean = df_clean.dropna(
        subset=["Lapangan_Pekerjaan"]
    )

    output_path = (
        f"{OUTPUT_DIR}/lapangan_pekerjaan_utama_{year}.csv"
    )

    df_clean.to_csv(
        output_path,
        index=False
    )

    print(df_clean.head())
```

```python
for year in [2024, 2025]:

    path = (
        f"{BPS_DIR}/PDRB/"
        f"PDRB Triwulanan Atas Dasar Harga Konstan (2010=100) "
        f"Menurut Lapangan Usaha di Kabupaten_Kota Seluruh Indonesia, {year}.csv"
    )

    df = load_csv(path)

    df_clean = df.iloc[3:].copy()

    df_clean.columns = (
        ["Kabupaten_Kota"] +
        [f"Sektor_{i}" for i in range(1, df.shape[1])]
    )

    df_clean = df_clean.dropna(
        subset=["Kabupaten_Kota"]
    )

    output_path = (
        f"{OUTPUT_DIR}/pdrb_wilayah_{year}.csv"
    )

    df_clean.to_csv(
        output_path,
        index=False
    )

    print(df_clean.head())
```
